```python
import seaborn as sns
#https://seaborn.pydata.org/examples/index.html 
# Go to the url above to see more examples of seaborn plots. You can also search for seaborn plots on google and you will find many examples of seaborn plots.
```

```python
sns.get_dataset_names() #Gives a list of all the datasets available in seaborn
```

```python
tips = sns.load_dataset("tips")
iris = sns.load_dataset("iris")
titanic = sns.load_dataset("titanic")
planets = sns.load_dataset("planets")   
```

```python
tips #shows the first and last 5 rows of the dataset
```

```python
#   SCATTER PLOT 

sns.scatterplot(x="tip",y="total_bill",data=tips,hue="day",size="size") #scatter plot of tips vs total bill
                                                            #hue is used to color the points based on the day of the week
                                                            #size is used to vary the size of the points based on the size column
```

```python
# HISTOGRAMS

sns.histplot(tips["tip"],kde=True,bins=20) #histogram of tips with kernel density estimate and 20 bins 
                                           #kde is used to plot the kernel density estimate which is a
                                           # smooth curve that represents the distribution of the data
```

```python
# DISTRIBUTION PLOT

sns.displot(tips["tip"],kde=True,bins=20) #distribution plot of tips with kernel density estimate and 20 bins
#Hist plot and dis plot are similar but dis plot is more flexible and can be used to plot multiple distributions at once
```

```python
# BARPLOTS
sns.barplot(x="sex",y="tip", data=tips, palette="YlGnBu") #bar plot of tips vs sex with a color palette of yellow to green to blue
```

```python
# BOXPLOTS
sns.boxplot(x="day",y="tip",data=tips,hue="sex",palette="YlGnBu")
#box plot of tips vs day with a color palette of yellow to green to blue

# Box Plot:
# - Middle line inside the box = Median (middle value)
# - Bottom of box = Q1 (25th percentile)
# - Top of box = Q3 (75th percentile)
# - Box = Middle 50% of the data (IQR = Q3 - Q1)
# - Whiskers = Normal minimum and maximum values
# - Circles outside the whiskers = Outliers
# - Taller box = More spread/variation in the data
# - Higher median = Generally higher values
```

```python
#STRIP PLOT
sns.stripplot(x="day",y="tip",data=tips, hue="sex",palette="YlGnBu",dodge=True) 
#strip plot of tips vs day with a color palette of yellow to green to blue
#dodge is used to separate the points based on the hue variable (sex)
# Here it seprated male and female tips.
```

```python
# JOINT PLOT
sns.jointplot(x="tip",y="total_bill",data=tips,kind="hex",cmap="YlGnBu")
#kind="hex" is used to plot a hexbin plot which is a bivariate histogram that shows the relationship between two variables
# join plot is used to plot the relationship between two variables along with their marginal distributions
# it uses both scatter plot and histograms to show the relationship between two variables
# cmap is used to specify the color map for the hexbin plot
```

```python
sns.jointplot(x="tip",y="total_bill",data=tips,kind="kde", shade=True,cmap="YlGnBu")
# kind="kde" is used to plot a kernel density estimate which is a smooth curve that represents the distribution of the data
# shade=True is used to fill the area under the curve with color
```

```python
 # Pair Plot

sns.pairplot(titanic.select_dtypes(['number']),hue="pclass") #pair plot of titanic with a color palette of yellow to green to blue
# PAIR PLOT — What is it?
# A pair plot shows relationships between ALL numeric columns at once
# It creates a grid of charts — each column vs every other column

# HOW TO READ IT:
# - Diagonal cells      → Distribution of that single column (histogram/KDE)
# - Off-diagonal cells  → Scatter plot between two columns
# - Color (hue)         → Different categories (here: pclass 1, 2, 3)

# WHAT THIS SPECIFIC PLOT SHOWS:
# - survived vs age     → Did age affect survival?
# - fare vs pclass      → Did ticket class affect fare price?
# - Each color          → 1st class (pink), 2nd class (purple), 3rd class (dark)
```

```python
# Select only numeric columns before correlation
titanic.select_dtypes(include='number').corr()
```

```python
# HEATMAPS

sns.heatmap(titanic.select_dtypes(include='number').corr(),annot=True,cmap="coolwarm") 
# heatmap of titanic with a color palette of cool to warm
#annot=True is used to annotate the heatmap with the correlation values
# Positive correlation = 1 → Both variables move in the same direction
# Negative correlation = -1 → Both variables move in opposite directions
# high correlation = 0.7 → Strong relationship between the two variables
# low correlation = 0.3 → Weak relationship between the two variables
# negative correlation = -0.7 → Strong relationship between the two variables but in opposite directions
# positive correlation = 0.7 → Strong relationship between the two variables in the same direction
```
